```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales = pd.read_csv('sales_data.csv')
sales.head()
```

```python
s = sales.head()
sns.barplot(data=sales, x= "Product_ID", y="Sales_Amount",hue="Region")
```

```python
sns.boxplot(data=s,
            x= "Product_Category",y="Unit_Cost")
```

```python
sns.violinplot(data=sales,
            x="Product_Category",y="Unit_Cost",
            hue="Region",split=True)
```

```python
sns.countplot(data=sales,
            x="Product_ID")
```
