```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales = pd.read_csv('sales_data.csv')
sales.head()
```

```python
sns.regplot(data=sales,
            x="Quantity_Sold",
            y="Unit_Price")
```

```python
sns.lmplot(
    data=sales,
    x="Quantity_Sold",
    y="Unit_Price",
    hue="Region"
)
plt.show()
```
