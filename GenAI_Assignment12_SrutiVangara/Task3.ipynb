```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pyplot
import seaborn as sns
```

```python
sales = pd.read_csv('sales_data.csv')
sales.head()
```

```python
sns.histplot(data = sales,
             x="Sales_Amount",
             hue= "Region")
```

```python
sns.displot(data = sales,
             x="Sales_Amount",
             hue= "Region",
             kind="hist")
```

```python
sns.kdeplot(data=sales,
            x="Sales_Amount",
            hue = "Region",
            fill = True)
```

```python
sns.rugplot(data=sales,
            x="Sales_Amount",
            hue = "Region")
```

```python
sns.histplot(data=sales,
              x="Sales_Amount", 
              kde=True, 
              bins=5 )
```
