```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales = pd.read_csv('sales_data.csv')
sales.head()
```

```python
g = sns.FacetGrid(data=sales, col="Region")
g.map(sns.scatterplot, "Quantity_Sold", "Unit_Price")
```

```python
sns.relplot(data=sales,
            x="Quantity_Sold",
            y="Unit_Cost",
            hue="Region",
            col= "Region")
```

```python
sns.catplot(data=sales,
            x="Product_Category",
            y="Sales_Amount",
            hue="Region",
            kind="bar")
```

```python
sns.displot(data=sales,
            x="Sales_Amount",
            hue="Region",
            kde=True)
```
